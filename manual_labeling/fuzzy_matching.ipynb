```python
import pandas as pd
from collections import defaultdict
from fuzzywuzzy import process, fuzz
```

```python
def check_for_duplicate_writers(csv_file_path, similarity_threshold=90):
    # Load the merged CSV file into a dataframe
    df = pd.read_csv(csv_file_path)

    # Extract the writer names and remove duplicates (case-insensitive)
    writer_names = df['Writer Name (English)'].dropna().unique()

    # Create a dictionary to store potential duplicates
    potential_duplicates = defaultdict(list)

    # Compare each writer name to every other name using fuzzy matching
    for writer in writer_names:
        # Find writer names that are similar based on the threshold
        matches = process.extractBests(writer, writer_names, scorer=fuzz.token_sort_ratio, limit=len(writer_names))
        
        # Store names that are similar enough to be considered duplicates
        for match_name, similarity in matches:
            if similarity >= similarity_threshold and writer != match_name:
                potential_duplicates[writer].append((match_name, similarity))

    # Print potential duplicates
    if potential_duplicates:
        print("Potential duplicate writer names found:")
        for writer, matches in potential_duplicates.items():
            print(f"\nWriter: {writer}")
            for match_name, similarity in matches:
                print(f"  - Possible match: {match_name} (Similarity: {similarity}%)")
    else:
        print("No potential duplicate writer names found.")
    
    return potential_duplicates

# Example usage
csv_file_path = '/Users/hamza/Research/Muharaf/merged_writer_1.csv'
check_for_duplicate_writers(csv_file_path, similarity_threshold=85)
```
